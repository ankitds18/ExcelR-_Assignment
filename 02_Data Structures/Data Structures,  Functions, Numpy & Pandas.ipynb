### 1. Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
numbers = [3, 5, 8, 2, 1]

numbers.append(10)
print("Append:", numbers)

numbers.extend([7, 9])
print("Extend:", numbers)

numbers.insert(1, 20)
print("Insert:", numbers)

numbers.remove(2)
print("Remove:", numbers)

numbers.pop()
print("Pop:", numbers)

print("Index of 8:", numbers.index(8))

print("Count of 2:", numbers.count(2))

numbers.sort()
print("Sort:", numbers)

numbers.reverse()
print("Reverse:", numbers)

numbers.clear()
print("Clear:", numbers)
```

# Section C: Exploring Pandas

### 11. Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:

- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ankit", "Rahul", "Aman", "Ravi", "Neha",
             "Priya", "Karan", "Sahil", "Pooja", "Ankit"],
    
    "Age": [21, 22, 20, 23, 21, 24, 22, 25, 23, 21],
    
    "City": ["Patna", "Delhi", "Mumbai", "Patna", "Delhi",
             "Kolkata", "Mumbai", "Delhi", "Patna", "Patna"],
    
    "Salary": [25000, 30000, np.nan, 28000, 32000,
               35000, np.nan, 40000, 30000, 25000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df[["Name", "City"]])
```

```python
df = df.drop("City", axis=1)

print(df)
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Salary"].mean()
```

```python
df = df.drop_duplicates()

print(df)
```
