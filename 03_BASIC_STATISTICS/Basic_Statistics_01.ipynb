# Data Visualization :

### ● Objective:

To visualize the distribution and relationship of numerical and categorical variables in the dataset.

```python
num_cols = df.select_dtypes(include = ['int','float']).columns
cat_cols = df.select_dtypes(include ='object').columns

print("Numerical Columns:\n", num_cols)
print("Categorical Columns:\n", cat_cols)
```

### Histogram :

Analyze the distribution :
skewness
presence of Outliers

# Boxplots :

```python
for col in num_cols: 
    plt.figure(figsize = (6,4))
    sns.histplot(df[col], bins = 30, kde = True, color = 'red')
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.show()
```

```python
for col in num_cols:
    plt.figure(figsize = (6,4))
    sns.boxplot(df[col])
    plt.title(col)
    plt.show()
```

# Bar Chart Analysis for Categorical Column :

```python
# Bar Charts for Date 
date_counts = df['Date'].value_counts()
plt.figure(figsize = (6,4))
plt.bar(date_counts.index, date_counts.values)
plt.xlabel("Date")
plt.ylabel("Count")
plt.xticks (rotation = 45)
plt.title(f"Date Analysis: ")
plt.show()
```

### Interpretation:

The dataset contains an equal number of transactions every day.
There are no missing dates.
No date dominates the dataset.

```python
# Bar Charts for Day :
plt.figure(figsize = (6,4))
day_counts = df['Day'].value_counts()
plt.bar(day_counts.index, day_counts.values)
plt.xlabel("Day")
plt.ylabel("Counts")
plt.title("Day Analysis")
plt.xticks(rotation = 45)
plt.show()
```

### Interpretation:

Sales are higher on Thursday.
Every other weekday sales are equal.

```python
# Bar Charts for SKU:
plt.figure(figsize = (15,4))
sku_counts = df['SKU'].value_counts()
plt.bar(sku_counts.index, sku_counts.values)
plt.xlabel("SKU")
plt.ylabel("Counts")
plt.title("SKU Analysis")
plt.xticks(rotation = 45)
plt.show()
```

### Interpretation :

All SKU are normally distributed.
There is no Null Or Missing value.

```python
# Bar Charts for City 
plt.figure(figsize = (6,4))
city = df['City'].value_counts()
plt.bar(city.index , city.values)
plt.xlabel("City")
plt.ylabel("Counts")
plt.xticks(rotation = 45)
plt.show()
```

### Interpretation :

Only one city (C) exist.
There is no variation in the City column.
It has only one unique value.

```python
# Bar chart for Brand 
plt.figure(figsize =(6,4))
brand = df['Brand'].value_counts()
plt.bar(brand.index , brand.values)
plt.xlabel("Brand")
plt.ylabel("Counts")
plt.title("Brand Analysis")
plt.xticks(rotation = 45)
plt.show()
```

### Interpretation :
